```python
import pandas as pd
from pyliftover import LiftOver
from alphagenome import colab_utils
from alphagenome.data import genome
from alphagenome.models import dna_client
from alphagenome.models import variant_scorers

# --- 1. 初始化模型和评分器 ---
dna_model = dna_client.create(colab_utils.get_api_key())
splicing_scorers = [
    variant_scorers.RECOMMENDED_VARIANT_SCORERS['SPLICE_SITES'],
    variant_scorers.RECOMMENDED_VARIANT_SCORERS['SPLICE_SITE_USAGE'],
    variant_scorers.RECOMMENDED_VARIANT_SCORERS['SPLICE_JUNCTIONS'],
]

for scorer in splicing_scorers:
  print(f'{scorer.name} (signed={scorer.is_signed})')

# --- 2. 保持不变的核心计算逻辑 ---
def compute_merged_splicing_score(df: pd.DataFrame) -> pd.DataFrame:
  """你原有的计算逻辑，保持完全不变"""
  df['variant_id'] = df['variant_id'].map(str)
  max_scores = (
      df.groupby(['variant_id', 'output_type'])['raw_score']
      .max()
      .reset_index()
      .pivot(index='variant_id', columns='output_type', values='raw_score')
      .fillna(0.0)
  )
  max_scores['alphagenome_splicing'] = (
      max_scores.get('SPLICE_SITES', 0.0)
      + max_scores.get('SPLICE_SITE_USAGE', 0.0)
      + max_scores.get('SPLICE_JUNCTIONS', 0.0) / 5.0
  )
  return max_scores.reset_index()

# --- 3. 新增的带 Liftover 的批量处理函数 ---
def process_variants_from_csv_hg19(input_csv_path: str, output_csv_path: str) -> pd.DataFrame:
    # 你的数据看起来是制表符分隔的，如果你的文件是纯逗号分隔的 csv，请把 sep='\t' 改为 sep=','
    df_input = pd.read_csv(input_csv_path, sep=',')

    # 初始化 Liftover 工具 (首次运行会自动从 UCSC 下载链文件，约需几秒钟)
    lo = LiftOver('hg19', 'hg38')

    # 新建两列来存储结果，保持原有 DataFrame 结构不变
    df_input['hg38_pos'] = None
    df_input['alphagenome_splicing'] = None

    print(f"正在处理 {len(df_input)} 个变异位点 (包含 hg19->hg38 转换)...")

    # 逐行遍历，完美保留重复项和原始顺序
    for index, row in df_input.iterrows():
        chrom = str(row['Chr'])
        pos_hg19 = int(row['Begin'])
        ref = str(row['Ref'])
        alt = str(row['Alt'])

        # --- A. 坐标转换 (Liftover) ---
        # pyliftover 使用 0-based 坐标，故输入 pos_hg19 - 1
        lifted = lo.convert_coordinate(chrom, pos_hg19 - 1)

        if not lifted:
            print(f"第 {index+1} 行: {chrom}:{pos_hg19} 转换失败 (可能位于 hg38 中被删除的区域)，跳过打分。")
            continue

        # 转换成功，将 0-based 坐标转回 1-based
        pos_hg38 = lifted[0][1] + 1
        df_input.at[index, 'hg38_pos'] = int(pos_hg38)

        # --- B. AlphaGenome 预测 ---
        variant = genome.Variant(
            chromosome=chrom,
            position=int(pos_hg38), # 注意：这里喂给模型的是转换后的 hg38 坐标
            reference_bases=ref,
            alternate_bases=alt,
        )

        interval = variant.reference_interval.resize(dna_client.SEQUENCE_LENGTH_1MB)

        scores = dna_model.score_variant(
            interval=interval,
            variant=variant,
            variant_scorers=splicing_scorers,
            organism=dna_client.Organism.HOMO_SAPIENS,
        )

        # --- C. 计算得分并填回原表 ---
        # 将当前的单一打分转换为 tidy dataframe
        df_single_score = variant_scorers.tidy_scores([scores])

        # 调用你定义的函数计算合并分数
        merged_result = compute_merged_splicing_score(df_single_score)

        # 提取计算好的分数，直接赋给当前行
        if not merged_result.empty:
            final_score = merged_result['alphagenome_splicing'].iloc[0]
            df_input.at[index, 'alphagenome_splicing'] = final_score

    # 保存最终结果 (包含原始的所有列，外加 hg38_pos 和最终得分)
    df_input.to_csv(output_csv_path, index=False, sep=',')
    print(f"处理完成！结果已完美保留原顺序，并保存至 {output_csv_path}")

    return df_input

# --- 4. 运行示例 ---
# 请确保你的输入文件表头包含: Chr, Begin, End, Ref, Alt
final_data = process_variants_from_csv_hg19('your_input_hg19.csv', 'predicted_scores.csv')
display(final_data)
```

```python
import os
import pandas as pd
# 如果在 Colab 运行，强烈建议挂载 Google Drive 以永久保存数据
from google.colab import drive
drive.mount('/content/drive')

from alphagenome import colab_utils
from alphagenome.data import genome
from alphagenome.models import dna_client
from alphagenome.models import variant_scorers

# --- 1. 初始化模型和评分器 ---
dna_model = dna_client.create(colab_utils.get_api_key())
splicing_scorers = [
    variant_scorers.RECOMMENDED_VARIANT_SCORERS['SPLICE_SITES'],
    variant_scorers.RECOMMENDED_VARIANT_SCORERS['SPLICE_SITE_USAGE'],
    variant_scorers.RECOMMENDED_VARIANT_SCORERS['SPLICE_JUNCTIONS'],
]

print("已加载的评分器:")
for scorer in splicing_scorers:
    print(f'- {scorer.name} (signed={scorer.is_signed})')

# --- 2. 保持不变的核心计算逻辑 ---
def compute_merged_splicing_score(df: pd.DataFrame) -> pd.DataFrame:
    """计算合并的剪接得分"""
    df['variant_id'] = df['variant_id'].map(str)
    max_scores = (
        df.groupby(['variant_id', 'output_type'])['raw_score']
        .max()
        .reset_index()
        .pivot(index='variant_id', columns='output_type', values='raw_score')
        .fillna(0.0)
    )
    max_scores['alphagenome_splicing'] = (
        max_scores.get('SPLICE_SITES', 0.0)
        + max_scores.get('SPLICE_SITE_USAGE', 0.0)
        + max_scores.get('SPLICE_JUNCTIONS', 0.0) / 5.0
    )
    return max_scores.reset_index()

# --- 3. 带有断点续传和定期保存的批量处理函数 ---
def process_variants_robust(input_csv_path: str, output_csv_path: str, save_interval: int = 100) -> pd.DataFrame:
    """
    input_csv_path: 原始输入文件路径
    output_csv_path: 结果保存路径（建议放在 Google Drive）
    save_interval: 每处理多少行保存一次（默认 100 行）
    """
    # 1. 断点续传逻辑：检查是否已经存在输出文件
    if os.path.exists(output_csv_path):
        print(f"\n[断点续传] 发现已存在的结果文件，正在读取: {output_csv_path}")
        df_data = pd.read_csv(output_csv_path, sep=',')
    else:
        print(f"\n[全新开始] 未发现历史记录，读取输入文件: {input_csv_path}")
        df_data = pd.read_csv(input_csv_path, sep=',')
        # 如果没有结果列，初始化为空
        if 'alphagenome_splicing' not in df_data.columns:
            df_data['alphagenome_splicing'] = None

    # 2. 筛选出还未处理的行（得分为空）
    unprocessed_mask = pd.isna(df_data['alphagenome_splicing'])
    unprocessed_indices = df_data[unprocessed_mask].index
    total_rows = len(df_data)
    remaining_rows = len(unprocessed_indices)

    print(f"总计 {total_rows} 个变异位点，当前剩余 {remaining_rows} 个待处理...")

    if remaining_rows == 0:
        print("所有位点已处理完毕！")
        return df_data

    # 3. 逐行处理剩余位点
    count = 0
    for index in unprocessed_indices:
        row = df_data.loc[index]

        chrom = str(row['chr'])
        pos = int(row['pos'])
        ref = str(row['ref'])
        alt = str(row['alt'])

        try:
            # --- AlphaGenome 预测 ---
            variant = genome.Variant(
                chromosome=chrom,
                position=pos,
                reference_bases=ref,
                alternate_bases=alt
            )
            interval = variant.reference_interval.resize(dna_client.SEQUENCE_LENGTH_1MB)
            scores = dna_model.score_variant(
                interval=interval,
                variant=variant,
                variant_scorers=splicing_scorers,
                organism=dna_client.Organism.HOMO_SAPIENS
            )

            # --- 计算得分并写入 DataFrame ---
            df_single_score = variant_scorers.tidy_scores([scores])
            merged_result = compute_merged_splicing_score(df_single_score)

            if not merged_result.empty:
                df_data.at[index, 'alphagenome_splicing'] = merged_result['alphagenome_splicing'].iloc[0]

        except Exception as e:
            # 捕获异常，防止某一行数据错误导致整个程序崩溃
            print(f"警告: 第 {index} 行 ({chrom}:{pos}) 处理失败，跳过。错误信息: {e}")
            continue

        # 4. 定期保存逻辑
        count += 1
        if count % save_interval == 0:
            print(f"已处理 {count} 个新位点，正在自动保存进度...")
            df_data.to_csv(output_csv_path, index=False, sep=',')

    # 5. 循环结束后做最后一次全量保存
    df_data.to_csv(output_csv_path, index=False, sep=',')
    print(f"\n✅ 全部处理完成！最终结果已保存至 {output_csv_path}")

    return df_data

# --- 4. 运行示例 ---
# 【重要】请将路径替换为你自己的 Google Drive 路径
# 确保你的输入文件表头为: chr, pos, ref, alt

INPUT_FILE = '/content/drive/MyDrive/clinvar+genomAD_splice.csv'     # 你的原始数据路径
OUTPUT_FILE = '/content/drive/MyDrive/predicted_scores.csv' # 你的结果保存路径

# 开始运行 (每处理 100 个位点保存一次，可根据需求修改 save_interval)
final_data = process_variants_robust(INPUT_FILE, OUTPUT_FILE, save_interval=100)

# 查看前几行结果
display(final_data.head())
```
